#### Merging registrations and employment datasets

```python
msme_data = pd.merge(registrations,employment, on="State_UT")
```

```python
msme_data.head()
```

#### Checking for mismatch in avg. no. of employees per MSME

```python
msme_data["Avg_Emp_Per_MSME"] = msme_data["Total_Employment"]/ msme_data["Total_MSME_Registrations"]
```

```python
msme_data["Emp_Mismatch"] = msme_data["Avg_Emp_Per_MSME"] != msme_data["Average_Employment_Per_MSME"]
```

```python
msme_data["Emp_Mismatch"].unique()
```

```python
msme_data = msme_data.drop(columns=["Emp_Mismatch","Avg_Emp_Per_MSME"], axis=1)
```

```python
msme_data.head()
```

#### Merging credit with registrations + employment to create the final analysis dataset

```python
msme_analysis = pd.merge(msme_data, credit, on= "State_UT")
```

```python
msme_analysis.head()
```

```python
msme_analysis.info()
```

```python
msme_analysis.describe()
```

#### Categorizing MSMEs as Leader, Credit gap or Employment gap

```python
conditions = [
    ((msme_analysis["Total_MSME_Registrations"] >= msme_analysis["Total_MSME_Registrations"].mean()) &
    (msme_analysis["Average_Credit_Per_MSME_Lakh"] >= msme_analysis["Average_Credit_Per_MSME_Lakh"].mean()) &
    (msme_analysis["Average_Employment_Per_MSME"] >= msme_analysis["Average_Employment_Per_MSME"].mean())),

    msme_analysis["Average_Credit_Per_MSME_Lakh"] < msme_analysis["Average_Credit_Per_MSME_Lakh"].mean(),

    msme_analysis["Average_Employment_Per_MSME"] < msme_analysis["Average_Employment_Per_MSME"].mean()   
]

choices = [
    "MSME Leader",
    "Credit Gap",
    "Employment Gap"
]

msme_analysis["Key_Insight"] = np.select(conditions, choices, "Emerging Ecosystem")
```

```python
group = msme_analysis.groupby("Key_Insight")
```

```python
group["State_UT"].count()
```

#### 17. More than 50% of the states have a credit gap.
#### 18. Around 16.67% of states have employment gap.

#### 19. MSME leader states

```python
msme_analysis[msme_analysis["Key_Insight"]=="MSME Leader"]
```

```python
msme_analysis.sort_values(by=['Average_Credit_Per_MSME_Lakh'], ascending=False).loc[:,['State_UT','Average_Credit_Per_MSME_Lakh','Average_Employment_Per_MSME']].head(10)
```

#### 20. More credit per MSME does not correspond to more employment per MSME

#### Exporting the final analysis into a csv file

```python
output_file = "MSME_India_Final_Analysis.csv"

msme_analysis.to_csv(output_file, index=False)
```
